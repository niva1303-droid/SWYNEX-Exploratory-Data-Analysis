**8. IQR Calculation**

```python
columns = ["Age", "Visit per week", "Avg time in gym"]

for column in columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ]

    print(f"\n{column}")
    print("Lower Bound:", lower_bound)
    print("Upper Bound:", upper_bound)
    print("Number of Outliers:", len(outliers))
```

```python
weekly_visit_outliers = df[df["Visit per week"] > 4.5]

print("\nWeekly Visit Outlier Values:")
print(weekly_visit_outliers["Visit per week"].value_counts())
```

## **Insights:**

1. Membership types are almost evenly distributed. Standard membership accounts for 507 members (50.7%), while Premium accounts for 493 members (49.3%). The difference is only 14 members, indicating a very balanced membership mix.

2. Weekly visit frequency is nearly identical across membership types. Premium members average 2.684 visits per week, while Standard members average 2.680 visits. The difference is only about 0.003 visits per week, so the dataset does not show a meaningful difference in visit frequency between the two membership types.

3. Average time spent in the gym is also very similar. Standard members spend an average of 105.60 minutes, compared with 104.91 minutes for Premium members. The difference is less than one minute, suggesting membership type has little observable relationship with workout duration in this dataset.

4. Additional service usage is highly balanced. Personal training has the highest participation at 51.8%, followed by group lessons at 50.3% and sauna usage at 49.3%. All three are close to 50%, so no single additional service strongly dominates.

5. Members are spread across different age groups, with ages 26–45 forming the largest segment. The 36–45 group represents 27.0%, closely followed by 26–35 at 26.8%. Combined, members aged 26–45 account for 53.8% of the dataset. Members under 18 represent 14.5%, while those aged 46+ account for 10.2%.

6. Age has virtually no linear relationship with gym duration. The correlation between age and average time spent in the gym is -0.041, indicating virtually no linear association between these variables in this dataset.

**Anomaly Analysis:** The IQR method identified no outliers for Age or Average Time in Gym. However, 124 records were identified as statistical outliers for weekly gym visits because they recorded 5 visits per week, exceeding the IQR upper bound of 4.5. Since 5 visits per week is a realistic and valid frequency, these observations were retained rather than treated as data errors.

## **Conclusion:**

The exploratory data analysis of the Gym Membership Dataset revealed a balanced distribution between Standard and Premium memberships, with very similar weekly visit frequencies and average gym durations across both membership types.

Members aged 26–45 form the largest age segment, accounting for 53.8% of the dataset. Additional services such as personal training, group lessons, and sauna usage are also evenly utilized, with approximately half of the members using each service.

The correlation between age and average time spent in the gym was -0.041, indicating virtually no linear relationship between the two variables.

The IQR method identified 124 statistical outliers in weekly gym visits. These records correspond to members visiting five times per week. Since this is a realistic gym attendance frequency, the records were retained and were not considered data-quality errors.

Overall, the analysis identified useful patterns in member demographics, membership types, gym attendance, service usage, and workout behavior while also distinguishing statistical outliers from potentially invalid data.
